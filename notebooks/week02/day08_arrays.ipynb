```python
import time

rng = np.random.default_rng(42)
v = rng.random(1_000_000)

def sum_sq_np(arr):
    return np.sum(arr ** 2)

def sum_sq_loop(arr):
    return sum(x * x for x in arr)

t = time.perf_counter(); print(sum_sq_np(v)); print("np  ", time.perf_counter() - t)
t = time.perf_counter(); print(sum_sq_loop(v)); print("loop", time.perf_counter() - t)
```

```python
print((0.1 + 0.2) + 0.3)
print(0.1 + (0.2 + 0.3))
print(np.isclose(sum_sq_np(v), sum_sq_loop(v)))
```
